```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../data/heart_disease_processed.csv")

print("Shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="target", data=df)

plt.title("Target Variable Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
df.hist(figsize=(15,12))

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="target",
    y="age",
    data=df
)

plt.title("Age vs Heart Disease")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="target",
    y="thalach",
    data=df
)

plt.title("Maximum Heart Rate vs Heart Disease")

plt.show()
```

## Key Findings

1. The dataset contains 303 observations and 14 columns.
2. No missing values were found after preprocessing.
3. The target classes are reasonably balanced.
4. Age shows a relationship with heart disease occurrence.
5. Maximum heart rate (thalach) appears to differ across target groups.
6. Correlation analysis indicates several clinical attributes are associated with heart disease prediction.
